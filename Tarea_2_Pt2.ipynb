```python
print('[INICIO 35/37] Cerrar la selección con validación y comparar los cuatro modelos en test.', flush=True)


def graficar_resultados_test(filas, titulo, detalle, incluir_ap=False):
    """Muestra métricas de detección y tiempo activo para las filas disponibles."""
    import matplotlib.pyplot as plt
    from matplotlib.ticker import FuncFormatter

    etiquetas = {
        'cnn_ssd': 'CNN + SSD',
        'yolo_scratch': 'YOLO11n\ndesde cero',
        'yolo_coco': 'YOLO11n\nCOCO',
        'fasterrcnn_coco': 'Faster R-CNN\nCOCO',
    }
    paneles = [
        ('Detección al umbral elegido', 'Porcentaje (%)', 100,
         [('precision', 'Precisión'), ('recall', 'Recall'), ('F1', 'F1')]),
        ('Entrenamiento activo', 'Minutos', 1 / 60, [('train_segundos', 'Tiempo')]),
    ]
    if incluir_ap:
        paneles.insert(0, ('Calidad de detección', 'Porcentaje (%)', 100,
                           [('AP50', 'AP50'), ('mAP50_95', 'mAP50–95')]))

    fig, ejes = plt.subplots(1, len(paneles), figsize=(4.8 * len(paneles), 4.6), squeeze=False)
    posiciones = range(len(filas))
    modelos = [etiquetas.get(fila['modelo'], fila['modelo']) for fila in filas]
    colores = ('#24548A', '#B34337', '#43776D')
    for eje, (nombre, unidad, factor, series) in zip(ejes[0], paneles, strict=True):
        ancho = 0.73 / len(series)
        maximo = 0.0
        for i, (clave, etiqueta) in enumerate(series):
            datos = [fila.get(clave) for fila in filas]
            valores = [valor * factor if valor is not None else float('nan') for valor in datos]
            maximo = max(maximo, max((valor for valor in valores if valor == valor), default=0))
            ubicaciones = [x + (i - (len(series) - 1) / 2) * ancho for x in posiciones]
            barras = eje.bar(ubicaciones, valores, ancho * 0.94, color=colores[i], label=etiqueta)
            decimales = 1 if clave == 'train_segundos' else 2
            eje.bar_label(barras, labels=[
                f'{valor:.{decimales}f}'.replace('.', ',') if dato is not None else ''
                for valor, dato in zip(valores, datos, strict=True)
            ], padding=3, fontsize=9)
        eje.set(title=nombre, ylabel=unidad, ylim=(0, maximo * 1.27 or 1))
        eje.set_xticks(list(posiciones), modelos)
        eje.yaxis.set_major_formatter(FuncFormatter(lambda valor, _: f'{valor:g}'.replace('.', ',')))
        eje.grid(axis='y', alpha=0.2)
        eje.set_axisbelow(True)
        eje.spines[['top', 'right']].set_visible(False)
        if len(series) > 1:
            eje.legend(frameon=False, loc='upper center', ncols=len(series), fontsize=8.5)

    fig.suptitle(titulo, fontsize=15, fontweight='bold')
    fig.text(0.5, 0.025, detalle, ha='center', fontsize=10, color='#46515D')
    fig.tight_layout(rect=(0, 0.075, 1, 0.92))
    display(fig)
    plt.close(fig)


if EJECUTAR_EVALUACION:
    if not auditoria or not manifiesto.get('partitions', {}).get('test') or not estadisticas:
        raise RuntimeError(
            'Para evaluar en test, usa la preparación real del notebook completo '
            'y los checkpoints de los cuatro modelos.'
        )
    # Reevaluar también las bases con el mismo dispositivo/software y batch 1.
    for nombre in ('cnn_ssd', 'yolo_scratch'):
        evaluar_validacion(nombre, auditoria, manifiesto, estadisticas, dispositivo=DISPOSITIVO_EJECUCION)
    cierre = cerrar_seleccion(auditoria, manifiesto, estadisticas)
    print('Ganador fijado con validación:', cierre['ganador_validacion'])
    for nombre in KINDS:
        evaluar_test(nombre, cierre, auditoria, manifiesto, estadisticas, dispositivo=DISPOSITIVO_EJECUCION)
    comparacion_test = comparar_disponibles(KINDS, 'test')
    if comparacion_test:
        graficar_resultados_test(
            comparacion_test, 'Comparación en test',
            'Métricas al umbral fijado en validación. El tiempo activo excluye validación y guardado.',
        )
else:
    print('Evaluación desactivada: no se cerró una selección ni se evaluó test en esta celda.', flush=True)
    # Test histórico: guion, Anexo A. Tiempos: training_summary.json de las corridas locales de 10 épocas.
    test_historico = [
        dict(modelo='cnn_ssd', precision=0.0067, recall=0.0656, F1=0.0121,
             AP50=0.0010, mAP50_95=0.0003, train_segundos=3452.2077339000098),
        dict(modelo='yolo_scratch', precision=0.0808, recall=0.0751, F1=0.0778,
             AP50=0.0257, mAP50_95=0.0090, train_segundos=3409.9467079999886),
    ]
    graficar_resultados_test(
        test_historico, 'Test histórico de los modelos desde cero',
        '10 épocas por modelo. El test común de los cuatro detectores sigue pendiente.',
        incluir_ap=True,
    )

# TAREA2_CELL_STATUS
print('[OK] Celda de código 35/37 lista. Celda de evaluación terminada; arriba se informa si se evaluó o si la opción estaba desactivada.', flush=True)
```
